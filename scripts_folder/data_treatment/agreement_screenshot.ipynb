```python
import os
import pathlib
import pytesseract
from PIL import Image
```

```python
pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
```

```python
root_path = pathlib.Path(os.getcwd())
root_path = root_path.parents[1]
```

```python
images = os.path.join(root_path, "data_folder", "agreement_screenshot")

all_text = ""
for image in os.listdir(images):
    image_path = os.path.join(images, image)
    open_image = Image.open(image_path)
    text = pytesseract.image_to_string(open_image)
    all_text += text

with open(os.path.join(root_path, "data_folder", "agreement_screenshot", "agreement.txt"), "w") as file:
    file.write(all_text)
```
